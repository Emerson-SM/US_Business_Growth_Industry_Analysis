```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
file_path = '../data/raw/Inc5000_Company_List_2014.csv'

df = pd.read_csv(file_path)
```

```python
df_clean = df.copy()
```

```python
print(df_clean.columns.tolist())
```

```python
#delete unnecessary columns 
columns_to_drop = [
    '_input', '_num', '_widgetName', '_source', '_resultNumber', '_pageUrl'
]

df_clean = df_clean.drop(columns=columns_to_drop)

print(df_clean.columns.tolist())
```

```python
#put workers 0 = nan
df_clean['workers'] = df_clean['workers'].replace(0, np.nan)

print(df_clean.info())
print(f'\n{df_clean.isnull().sum()}')
```

```python
# Verify dataset dimensions
print("Dataset shape:", df_clean.shape)

# Verify missing values
print("\nMissing values:")
print(df_clean.isnull().sum())

# Verify duplicated records
print("\nDuplicate rows:", df_clean.duplicated().sum())

# Verify duplicated IDs
print("Duplicate IDs:", df_clean['id'].duplicated().sum())
```

```python
#export the clean data
output_path = "../data/processed/inc5000_2014_clean.csv"

df_clean.to_csv(output_path, index=False)

print("Clean dataset exported successfully.")
```

```python
#comprobar data nueva existente

df_check = pd.read_csv(output_path)

print("Exported dataset shape:", df_check.shape)
print("\nMissing values:")
print(df_check.isnull().sum())
```
